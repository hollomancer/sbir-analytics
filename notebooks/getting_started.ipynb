# Getting started with SBIR Analytics notebooks

This deterministic smoke notebook verifies that the local project configuration can load. It is executed by CI and deliberately does not contact S3, HuggingFace, or other external services.

For research work, copy `_template.ipynb` or reuse a companion notebook under `examples/`. See `notebooks/README.md` for the notebook-first workflow.

```python
import os
from pathlib import Path

def find_repo_root(start: Path = Path.cwd()) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").exists() and (candidate / "sbir_etl").exists():
            return candidate
    raise RuntimeError("Run this notebook from inside the sbir-analytics checkout")


project_root = find_repo_root()
os.chdir(project_root)

from sbir_etl.config.loader import get_config
from sbir_etl.utils.logging_config import setup_logging

# Initialize environment
setup_logging()
config = get_config()

print(f"Loaded configuration for environment: {config.pipeline.get('environment', 'unknown')}")
print(f"Project root: {project_root}")
```

## Start a research exploration

The template captures the research question, decision, data contract, deterministic parameters, findings, caveats, and graduation checklist.

```python
template = project_root / "notebooks" / "_template.ipynb"
examples = sorted((project_root / "notebooks" / "examples").glob("*.ipynb"))
print(f"Template: {template.relative_to(project_root)}")
for example in examples:
    print(f"Example:  {example.relative_to(project_root)}")
```
